```python
import pandas as pd
import numpy as np

rng = np.random.default_rng(42)
n = 20

df = pd.DataFrame({
    "patient_id":  [f"P{i:03d}" for i in range(1, n + 1)],
    "age":         rng.integers(25, 80, size=n),
    "weight_kg":   rng.uniform(55, 110, size=n).round(1),
    "dose_mgy":    rng.uniform(10, 500, size=n).round(2),
    "organ":       rng.choice(["heart", "lung", "liver", "kidney"], size=n),
    "is_female":   rng.choice([True, False], size=n),
})

df.head()
```

```python
# Add a dose category column
df["dose_category"] = pd.cut(
    df["dose_mgy"],
    bins=[0, 100, 300, 500],
    labels=["low", "medium", "high"],
)

# Summary by organ
df.groupby("organ")[["dose_mgy", "age", "weight_kg"]].mean().round(2)
```
