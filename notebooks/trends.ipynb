# Dose trends over time

Median and IQR of DAP per month/quarter/year, per procedure.
Reads all years under the data folders when `YEAR = None`.

```python
%load_ext autoreload
%autoreload 2

import xa_dose_analysis as xa

xa.setup_logging("INFO")  # use "WARNING" for less output
cfg = xa.load_config()
```

```python
ANALYSIS = "pci"
YEAR = None  # None = all years in the data folders
FREQ = "Q"  # "M", "Q" or "Y"
```

```python
ds = xa.load_dataset(cfg, year=YEAR)
data = xa.select_analysis(ds, cfg, ANALYSIS)
```

```python
# Trend plot per configured procedure (with its DRL when configured):
for procedure in cfg.analysis(ANALYSIS).procedures:
    xa.plot_trend(data, procedure.name, freq=FREQ, drl=procedure.drl_dap, save=False)
```

```python
# The numbers behind one of the plots:
procedure = cfg.analysis(ANALYSIS).procedures[0].name
xa.trend_table(data, procedure, freq=FREQ)
```

```python
# Per-room trends for one procedure:
xa.trend_table(data, procedure, freq="Y", by_room=True)
```
