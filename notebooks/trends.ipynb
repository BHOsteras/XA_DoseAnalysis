# Dose trends over time

Median and IQR of DAP per month/quarter/year, per procedure —
for all labs together and per lab.
Reads all years under the data folders when `YEAR = None`.

```python
%load_ext autoreload
%autoreload 2

import xa_dose_analysis as xa

xa.setup_logging("INFO")  # use "WARNING" for less output
cfg = xa.load_config()
```

```python
ANALYSIS = "pci"
YEAR = None  # None = all years in the data folders
FREQ = "M"  # "M", "Q" or "Y"
```

```python
ds = xa.load_dataset(cfg, year=YEAR)
data = xa.select_analysis(ds, cfg, ANALYSIS)
```

## All labs together

```python
# Trend plot per configured procedure (with its DRL when configured):
for procedure in cfg.analysis(ANALYSIS).procedures:
    xa.plot_trend(data, procedure.name, freq=FREQ, drl=procedure.drl_dap, save=False)

# clip=False keeps the full range instead; y_max=... caps the axis by hand.
```

```python
PROCEDURE = "Coronary angiography"  # the procedure used by the per-lab cells below
#PROCEDURE = "PCI"  # the procedure used by the per-lab cells below
#PROCEDURE = "TAVI"  # the procedure used by the per-lab cells below
#PROCEDURE = "Mitraclip"  # the procedure used by the per-lab cells below
#PROCEDURE = "PTSMA"  # the procedure used by the per-lab cells below
#PROCEDURE = "PFO"  # the procedure used by the per-lab cells below

# The numbers behind one of the plots:
xa.trend_table(data, PROCEDURE, freq=FREQ)
```

## Per lab

Two views of the same numbers for one procedure (`PROCEDURE` above):

- **All labs in one plot** — one median line per lab, counts in the legend.
  No IQR bands: with several labs they overlap too much to read.
- **One plot per lab** — median and IQR as in the plots above, and the same
  y-axis on every figure so the labs can be compared by eye.

A single extreme period no longer sets that shared y-axis for every lab: the
axis is capped just above the bulk of the data, and the periods running past
the cap are marked above it with their count and value (a filled triangle when
the median itself is outside, a hollow one when only the IQR band is).

```python
#PROCEDURE = "Coronary angiography"  # the procedure used by the per-lab cells below
#PROCEDURE = "PCI"  # the procedure used by the per-lab cells below
PROCEDURE = "TAVI"  # the procedure used by the per-lab cells below
```

```python
# All labs in one plot, one median line per lab:
proc = cfg.analysis(ANALYSIS).procedure(PROCEDURE)
xa.plot_trend(data, proc.name, freq=FREQ, drl=proc.drl_dap, by_room=True, save=False)

# For every configured procedure instead:
# for proc in cfg.analysis(ANALYSIS).procedures:
#     xa.plot_trend(data, proc.name, freq=FREQ, drl=proc.drl_dap, by_room=True, save=False)
```

```python
# One plot per lab, each with median and IQR:
proc = cfg.analysis(ANALYSIS).procedure(PROCEDURE)
figures = xa.plot_trend_per_room(data, proc.name, freq=FREQ, drl=proc.drl_dap, save=True)

# rooms=[...] limits which labs are plotted; share_y=False lets each figure autoscale.
# clip=False turns the capping off, so the tallest period sets the axis for all labs.
```

```python
# The numbers behind the per-lab plots:
xa.trend_table(data, PROCEDURE, freq="Y", by_room=True)
```
