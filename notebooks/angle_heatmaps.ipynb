# Air kerma by C-arm angle

Heatmaps of where the dose is delivered in the LAO/RAO x cranial/caudal angle plane,
from exposure-level DoseTrack exports.

The exposure-level files are large; the combined data is cached as a pickle file in the
data folder, so only the first run is slow.

```python
%load_ext autoreload
%autoreload 2

import xa_dose_analysis as xa

xa.setup_logging("INFO")  # use "WARNING" for less output
cfg = xa.load_config()
```

```python
ANALYSIS = "pci"
YEAR = 2025
BIN_SIZE = 10  # degrees
```

```python
# Procedure-level data for grouping the exposures by mapped procedure:
ds = xa.load_dataset(cfg, year=YEAR)
data = xa.select_analysis(ds, cfg, ANALYSIS)
```

```python
# Exposure-level data (cached after the first run):
exposures = xa.load_exposure_data(cfg, year=YEAR)
len(exposures)
```

```python
for procedure in cfg.analysis(ANALYSIS).procedures:
    proc_exposures = xa.filter_exposures_for_procedure(exposures, data, procedure.name)
    xa.plot_air_kerma_angle_heatmap(proc_exposures, procedure.name, bin_size=BIN_SIZE, save=False)
```
